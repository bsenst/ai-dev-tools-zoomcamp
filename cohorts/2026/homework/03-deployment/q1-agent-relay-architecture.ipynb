# Question 1: Agent Relay architecture

This notebook downloads the public Agent Relay starter files and checks the architecture directly from the project documentation and source code.

The expected answer is: **Agents claim tasks from a DB through an HTTP API.**

```python
from urllib.request import urlopen

REPO = "https://raw.githubusercontent.com/alexeygrigorev/agent-relay/main"

def fetch(filename):
    with urlopen(f"{REPO}/{filename}") as response:
        return response.read().decode("utf-8")

readme = fetch("README.md")
main = fetch("main.py")
database = fetch("database.py")
storage = fetch("storage.py")

print("Downloaded README.md, main.py, database.py, and storage.py")
```

```python
# The README identifies the service and its persistent queue.
assert "FastAPI service" in readme
assert "SQLite persists the queue and attempts" in readme
print("README evidence:")
for line in readme.splitlines():
    if "FastAPI service" in line or "SQLite persists" in line or "workers execute" in line:
        print("  " + line)
```

```python
# The API exposes agent registration and task operations.
assert "FastAPI" in main
assert "/api/v1/agents" in main
assert "claim" in main.lower()
assert "complete" in main.lower() or "result" in main.lower()
print("HTTP API evidence found: FastAPI, agent registration, task claiming, and result handling")
```

```python
# Persistence and delivery logic live in database.py and storage.py.
assert "sqlalchemy" in database.lower()
assert "sqlite" in database.lower()
storage_terms = storage.lower()
assert "claim" in storage_terms
assert "task" in storage_terms
print("Database evidence found: SQLite models plus storage operations for tasks and claims")
```

```python
answer = "Agents claim tasks from a DB through an HTTP API."
print(answer)
print("")
print("Why: clients communicate with the FastAPI service; the service persists the queue in SQLite; workers claim tasks through the API and submit results.")
```
